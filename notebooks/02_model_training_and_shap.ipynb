# 02. Random Forest Model Training & Feature Importance Analysis

Trains the `CementProcessModel` pipelines for both `Energy Consumption (kWh/t-clinker)` and `CO2 Emissions (kg/h)`, evaluates test-set $R^2$ / RMSE / CV scores, and inspects feature importances.

```python
from cement_opt.config import load_config
from cement_opt.data import get_controllable_feature_names, load_and_merge_datasets, prepare_modeling_dataframe
from cement_opt.models import CementProcessModel

cfg = load_config("../configs/default.yaml")
merged = load_and_merge_datasets(cfg.paths.energy_csv, cfg.paths.co2_csv)
df = prepare_modeling_dataframe(merged, include_fuel_encoding=cfg.data.include_fuel_encoding)
features = get_controllable_feature_names(include_fuel_encoding=cfg.data.include_fuel_encoding)

energy_model = CementProcessModel(cfg.data.energy_target, features, cfg.model, cfg.thermodynamics)
co2_model = CementProcessModel(cfg.data.co2_target, features, cfg.model, cfg.thermodynamics)

print(energy_model.train(df))
print(co2_model.train(df))
```

```python
print("Top Energy Drivers:")
display(energy_model.get_feature_importances().head(10))

print("Top CO2 Drivers:")
display(co2_model.get_feature_importances().head(10))
```
